# Data Cleaning & Structural Validation – Customer Churn
Clean and validate the supplied customer churn dataset using Pandas.

```python
import pandas as pd
df = pd.read_csv('customer_churn_sample.csv')
print('Shape:', df.shape)
display(df.head())
display(df.dtypes)
```

## Inspection findings
- Dataset size: **15 rows × 11 columns**.
- Missing values: **0**.
- Exact duplicate rows: **0**.
- No missing values or duplicate records were found in the supplied sample.

```python
display(df.isna().sum())
print('Duplicate rows:', df.duplicated().sum())
```

## Cleaning decisions
- **Column headers:** renamed to consistent lowercase `snake_case`.
- **Categorical strings:** leading/trailing whitespace removed.
- **Missing values:** none were present, so no imputation or deletion was performed.
- **Duplicates:** none were present, so no rows were removed.
- **Numeric types:** numeric fields were explicitly converted and validated.
- **Dates:** no date column exists in this sample, so date formatting was not applicable.

```python
rename_map = {
    'CustomerID':'customer_id', 'Gender':'gender', 'Age':'age',
    'TenureMonths':'tenure_months', 'SubscriptionType':'subscription_type',
    'MonthlyCharges':'monthly_charges', 'TotalCharges':'total_charges',
    'ContractType':'contract_type', 'SupportTickets':'support_tickets',
    'PaymentMethod':'payment_method', 'Churn':'churn'
}
clean = df.rename(columns=rename_map).copy()
for col in clean.select_dtypes(include='object').columns:
    clean[col] = clean[col].astype('string').str.strip()
numeric_cols = ['age','tenure_months','monthly_charges','total_charges','support_tickets']
for col in numeric_cols:
    clean[col] = pd.to_numeric(clean[col], errors='coerce')
for col in ['age','tenure_months','support_tickets']:
    clean[col] = clean[col].astype('Int64')
clean = clean.drop_duplicates().reset_index(drop=True)
display(clean.head())
```

## Structural validation
Check unique IDs, sensible numeric ranges, valid churn labels, and missing values.

```python
assert clean['customer_id'].is_unique
assert clean['age'].between(18, 100).all()
assert (clean['tenure_months'] >= 0).all()
assert ((clean['monthly_charges'] >= 0) & (clean['total_charges'] >= 0)).all()
assert (clean['support_tickets'] >= 0).all()
assert clean['churn'].isin(['Yes','No']).all()
assert clean.isna().sum().sum() == 0
print('All structural validation checks passed.')
display(clean.dtypes)
```

```python
clean.to_csv('customer_churn_cleaned.csv', index=False)
print('Cleaned CSV exported successfully.')
```
